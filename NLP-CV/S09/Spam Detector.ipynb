```python
from tkinter import *
import joblib
from torch.utils.hipify.hipify_python import preprocessor
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
ps=PorterStemmer()
swords=stopwords.words('english')
def clean_text(sent):
    tokens1=word_tokenize(sent)
    tokens2=[token for token in tokens1 if token.isalpha()]
    tokens3=[ps.stem(token.lower()) for token in tokens2 if token.lower() not in swords]
    return tokens3
```

```python
top=Tk()
top.title('SPAM Message Detector')
top.geometry('800x500')

classifier=joblib.load('classifier.model')
preprocessor=joblib.load('preprocessor.model')
def check():
    message=t.get('1.0',END)
    mes_new=preprocessor.transform([message])
    pred=classifier.predict(mes_new)
    if pred[0]=='ham':
        result.config(text='Message is NOT Spam!')
    else:
        result.config(text='Message is Spam!')
Label(top,text='Enter your meessage here').place(x=100,y=10)
t=Text(top)
t.place(x=100,y=40)
b=Button(top,text='Check Message',command=check)
b.place(x=200,y=450)
result=Label(top)
result.place(x=400,y=450)
top.mainloop()
```

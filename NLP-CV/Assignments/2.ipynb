```python
from nltk.tokenize import word_tokenize
from nltk import pos_tag, RegexpParser
```

```python
tags = pos_tag(word_tokenize(nasa_data))
```

```python
tags
```

```python
import string
without_punc = [i for i in word_tokenize(nasa_data) if not i in string.punctuation]
' '.join(without_punc)
```

```python
from nltk.corpus import stopwords
swords=stopwords.words('english')
```

```python
stp_rem = [i for i in word_tokenize(nasa_data) if not i in swords]
' '.join(stp_rem)
```

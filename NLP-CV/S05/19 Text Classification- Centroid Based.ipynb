### Define cleaning function

```python
import re,string
```

```python
def cleanResume(resumeText):
    resumeText=re.sub(r"http\S+\s",' ',resumeText) #URLs
    resumeText=re.sub(r"RT|CC",' ',resumeText)     #RT | cc
    resumeText=re.sub(r"#\S+",' ',resumeText)      #Hashtags
    resumeText=re.sub(r"@\S+",' ',resumeText)      #mentions and email-ids
    resumeText=re.sub(r"[%s]"%re.escape(string.punctuation),' ',resumeText)
    resumeText=re.sub(r"[^\x00-\x7f]",r' ',resumeText)  #keep only ASCII letters
    resumeText=re.sub(r"\s+",' ',resumeText)            #extra whitespace
    return resumeText
```

### TF*IDF Vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(stop_words='english')
```

```python
x_new=tfidf.fit_transform(cleaned)
```

```python
x_new.shape
```

```python
tfidf.get_feature_names_out()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_new,y,random_state=0,stratify=y)
```

```python
x_train.shape,x_test.shape
```

```python
from sklearn.neighbors import NearestCentroid
```

```python
nc=NearestCentroid()
```

```python
nc.fit(x_train,y_train)
```

```python
nc.classes_
```

### Performance Evaluation

```python
from sklearn.metrics import accuracy_score, classification_report
from sklearn.metrics import ConfusionMatrixDisplay
```

```python
y_pred=nc.predict(x_test)
```

```python
plt.figure(figsize=(16,16))
ConfusionMatrixDisplay.from_predictions(y_test,y_pred)
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

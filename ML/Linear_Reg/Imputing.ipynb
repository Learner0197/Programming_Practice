```python
concrete = pd.read_csv(r"Concrete_Data.csv")
X = concrete.drop("Strength", axis=1)
y = concrete["Strength"]

lr = LinearRegression()
lr.fit(X, y)
tst = pd.read_csv(r"testConcreteMissing.csv")

imputer = SimpleImputer(strategy="mean").set_output(transform="pandas")
imputer.fit(X)
tst = imputer.transform(tst)
print(tst.mean())
y_pred = lr.predict(tst)
print(y_pred)
```

```python
from sklearn.model_selection import train_test_split

chem = pd.read_csv("ChemicalProcess.csv")
X = chem.drop("Yield", axis=1)
y = chem["Yield"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25)
```

```python
imputer = SimpleImputer(strategy="mean").set_output(transform="pandas")
X_trn_imp = imputer.fit(X_train).transform(X_train)

X_tst_imp = imputer.transform(X_test)
```

```python
lr = LinearRegression()
lr.fit(X_trn_imp, y_train)
y_pred = lr.predict(X_tst_imp)
mean_absolute_error(y_test, y_pred)
```

```python
imputer = SimpleImputer(strategy="median").set_output(transform="pandas")
X_trn_imp = imputer.fit(X_train).transform(X_train)

X_tst_imp = imputer.transform(X_test)
```

```python
lr = LinearRegression()
lr.fit(X_trn_imp, y_train)
y_pred = lr.predict(X_tst_imp)
mean_absolute_error(y_test, y_pred)
```

#### Iterative Imputer

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
imputer = IterativeImputer(initial_strategy="mean")
X_trn_imp = imputer.fit_transform(X_train)
X_tst_imp = imputer.transform(X_test)
lr = LinearRegression()
lr.fit(X_trn_imp, y_train)
y_pred = lr.predict(X_tst_imp)
mean_absolute_error(y_test, y_pred)
```

```python
import pandas as pd
import numpy as np
from numpy.random import logistic
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, classification_report
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt
import warnings; warnings.filterwarnings('ignore')
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, StandardScaler, MinMaxScaler
```

```python
cancer = pd.read_csv(r"C:\Users\dai\PycharmProjects\ML\Logistic_reg\BreastCancer.csv")
X = cancer.drop(["Code", "Class"], axis=1)
y = cancer["Class"]
```

```python
def knn(X, y):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25, stratify=y)
    scores = []
    for n in range(1, 11):
        knn = KNeighborsClassifier(n_neighbors=n)
        knn.fit(X_train, y_train)
        y_pred = knn.predict(X_test)
        scores.append([accuracy_score(y_test, y_pred), n])

    scores = pd.DataFrame(scores, columns=["accuracy", "n"])
    scores.sort_values("accuracy", ascending=False)
    return scores
knn(X, y)
```

```python
hr = pd.read_csv(r"C:\Users\dai\PycharmProjects\ML\Logistic_reg\HR_comma_sep.csv")
hr = pd.get_dummies(hr, drop_first=True)
X = hr.drop(["left"], axis=1)
y = hr.left
```

```python
knn(X, y)
```

```python
ss = StandardScaler()
def knn(X, y, ss):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25, stratify=y)
    X_trn_ss = ss.fit_transform(X_train)
    X_tst_ss = ss.transform(X_test)
    scores = []
    for n in range(1, 11):
        knn = KNeighborsClassifier(n_neighbors=n)
        knn.fit(X_trn_ss, y_train)
        y_pred = knn.predict(X_tst_ss)
        scores.append([accuracy_score(y_test, y_pred), n])

    scores = pd.DataFrame(scores, columns=["accuracy", "n"])
    scores.sort_values("accuracy", ascending=False)
    return scores

knn(X, y, ss)
```

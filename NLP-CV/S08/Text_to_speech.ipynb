```python
generator=pipeline('text-generation',model='gpt2')
```

```python
data=generator('I want to learn Python Programming so',max_length=30,num_return_sequences=3)
```

### Conversational

```python
vqa=pipeline('visual-question-answering')
```

```python
from PIL import Image
```

```python
img=Image.open('cat.jpg')
```

```python
img
```

```python
ans=vqa(question='What is present in image?',image=img)
```

```python
ans
```

```python
ans=vqa(question='What is the color of cat?',image=img)
ans[0]['answer']
```

```python
ans=vqa(question='What cat is doing?',image=img)
ans[0]['answer']
```

```python
ans=vqa(question='What emotion is on the cat face?',image=img,topn=1)
ans[0]['answer']
```

### Text to speech

```python
pipe=pipeline('text-to-speech',model='suno/bark-small')
```

```python
text='मचाएंगे'
output=pipe(text)
```

```python
from IPython.display import Audio
Audio(output['audio'],rate=output['sampling_rate'])
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, r2_score
from sklearn.neighbors import KNeighborsRegressor
from sklearn.compose import make_column_transformer, make_column_selector
import warnings;

warnings.filterwarnings('ignore')
from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler, OneHotEncoder
```

```python
hr = pd.read_csv('HR_comma_sep.csv')
hr_tst = pd.read_csv('../Logistic_reg/tst_hr.csv')
```

```python
X = hr.drop(["left"], axis=1)
y = hr["left"]
cat_cols = list(X.columns[X.dtypes == "object"])
num_cols = list(X.columns[X.dtypes != "object"])
```

```python
def knn(X, y):
    # Encode categorical columns
    X = X.copy()
    for col in X.select_dtypes(include='object').columns:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col])

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25)
    scores = []
    for n in range(1, 11):
        knn = KNeighborsRegressor(n_neighbors=n)
        knn.fit(X_train, y_train)
        y_pred = knn.predict(X_test)
        scores.append([r2_score(y_test, y_pred), n])

    scores = pd.DataFrame(scores, columns=["r2_score", "n"])
    scores = scores.sort_values("r2_score", ascending=False)
    return scores

knn(X, y)
```

```python
def knn_scl(X, y, ss):
    # Encode categorical columns
    X = X.copy()
    for col in X.select_dtypes(include='object').columns:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col])

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=25) #Stratify is for classification and not for regression.
    X_trn_ss = ss.fit_transform(X_train)
    X_tst_ss = ss.transform(X_test)
    scores = []
    for n in range(1, 11):
        knn = KNeighborsRegressor(n_neighbors=n)
        knn.fit(X_trn_ss, y_train)
        y_pred = knn.predict(X_tst_ss)
        scores.append([r2_score(y_test, y_pred), n])

    scores = pd.DataFrame(scores, columns=["r2_score", "n"])
    scores = scores.sort_values("r2_score", ascending=False)
    return scores

ss = MinMaxScaler()
knn_scl(X, y, ss)
```

```python
ohe = OneHotEncoder(sparse_output=False, drop='first').set_output(transform='pandas')
trns = make_column_transformer((ohe, cat_cols), remainder='passthrough', verbose_feature_names_out=False).set_output(transform='pandas')
X_trns = trns.fit_transform(X)
X_trns
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_trns, y, test_size=0.3, random_state=25)
```

```python
ss = StandardScaler().set_output(transform='pandas')
knn = KNeighborsRegressor(n_neighbors=1)
X_train_scaled = ss.fit_transform(X_train)
X_test_scaled = ss.transform(X_test)
knn.fit(X_train_scaled, y_train)
```

```python
y_pred = knn.predict(X_test_scaled)
tst = pd.read_csv("../Logistic_reg/tst_hr.csv")
tst_trns = trns.transform(tst)
tst_scl = ss.transform(tst_trns)
print(knn.predict(tst_scl))
```

```python
r2_score(y_test, y_pred)
```

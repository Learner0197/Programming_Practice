```python
# Extract video metadata
width = int(vid.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(vid.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(vid.get(cv2.CAP_PROP_FRAME_COUNT))
fps = int(vid.get(cv2.CAP_PROP_FPS))
```

```python
print('Width: {}, Height: {}'.format(width, height))
print('Total Frames: {}'.format(total_frames))
print('FPS: {}'.format(fps))
```

```python
duration = total_frames / fps # total duration of the video
```

```python
# Display the video
vid = cv2.VideoCapture('images/nature.mp4')

while True:
    ret, frame = vid.read()
    if ret == False:
        break
    cv2.imshow('My Video', frame)
    if cv2.waitKey(5) == ord('q'):
        break

cv2.destroyAllWindows()
```

```python
cv2.imwrite(f"images/myframe.jpg", frame) #Store a particular frame
```

```python
# Display the video
vid = cv2.VideoCapture('images/nature.mp4')

frame_count=0
while True:
    ret, frame = vid.read()
    if ret == False:
        break
    cv2.imwrite(f"images/frames/myframe_{frame_count}.jpg", frame) #Store all frames in the video
    cv2.imshow('My Video', frame)
    frame_count += 1
    if cv2.waitKey(5) == ord('q'):
        break

cv2.destroyAllWindows()
```

```python
time = "00:00:45.00"
```

```python
hrs, mins, secs = time.split(':')
```

```python
hrs, mins, secs
```

```python
trigger_frame = fps * (int(hrs) * 3600 + int(mins) * 60 + float(secs))
```

```python
trigger_frame
```

```python
vid.set(1, trigger_frame) #Jump to the trigger frame
```

```python
ret, frame = vid.read()
cv2.imshow("Trigger Frame", frame)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

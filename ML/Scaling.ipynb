```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
arr1 = np.arange(400, 600, 20)
np.random.shuffle(arr1)
arr2 = np.linspace(0, 10, 10)
np.random.shuffle(arr2)
arr3 = np.linspace(0, 1, 10)
np.random.shuffle(arr3)

arr = np.array([arr1, arr2, arr3])
arr
```

```python
df = pd.DataFrame(arr)
df = df.transpose().rename({0:"X1", 1:"X2", 2:"X3"}, axis=1)
df
```

```python
ss = StandardScaler().set_output(transform="pandas")
ss.fit(df)
df1 = ss.transform(df)
df1
```

```python
df1.describe().loc[["mean", "std"],:]
```

```python
mm = MinMaxScaler().set_output(transform="pandas")
mm.fit(df)
df2 = mm.transform(df)
df2
```

```python
df2.describe().loc[["mean", "std"],:]
```

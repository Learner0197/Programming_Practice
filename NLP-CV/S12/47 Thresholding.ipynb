#### Binary Thresholding

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread('images/dog.jpg',0)
```

```python
for i in range(img.shape[0]-1):
    for j in range(img.shape[1]-1):
        if img[i,j]>=128:
            img[i,j]=255
        else:
            img[i,j]=0
```

```python
plt.imshow(img,cmap='gray')
```

```python
img=cv2.imread('images/lenna.png',0)
binary_img=img>120
plt.imshow(binary_img,cmap='gray')
```

```python
img=cv2.imread('images/lenna.png',0)
inv_binary_img=img<120
plt.imshow(inv_binary_img,cmap='gray')
```

```python
binary_img.dtype
```

```python
binary_img=binary_img.astype(np.uint8)*255
```

```python
binary_img.dtype
```

```python
plt.subplot(121)
plt.imshow(img,cmap='gray')
plt.subplot(122)
plt.imshow(binary_img,cmap='gray')
```

```python
thresh,new_img=cv2.threshold(img,127,255,cv2.THRESH_BINARY)
```

```python
plt.figure(figsize=[18,6])
plt.subplot(141)
plt.title('Original Image')
plt.imshow(img,cmap='gray')
plt.subplot(142)
plt.title('By array operations')
plt.imshow(binary_img,cmap='gray')
plt.subplot(143)
plt.title('By array operations(inverse)')
plt.imshow(inv_binary_img,cmap='gray')
plt.subplot(144)
plt.title('By functions')
plt.imshow(new_img,cmap='gray')
```

#### Averaging Kernel

```python
img=cv2.imread('images/dog.jpg')
```

```python
kernel=np.zeros([3,3])+1/9
```

```python
kernel
```

```python
blurred_image=cv2.filter2D(img,-1,kernel)
```

```python
cv2.imshow('blurred_image',blurred_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
kernel=np.zeros([9,9])+1/81
blurred_image=cv2.filter2D(img,-1,kernel)
blurred_image=np.hstack([img,blurred_image])
```

```python
cv2.imshow('blurred_image',blurred_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Gaussian Kernel

```python
kernel_size=7
kernel=cv2.getGaussianKernel(kernel_size,1.23)
```

```python
kernel
```

```python
gaussian_blurred_image=cv2.filter2D(img,-1,kernel)
gaussian_blurred_image=np.hstack([img,gaussian_blurred_image])
```

```python
cv2.imshow('Gaussian Blurred image',gaussian_blurred_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Median Blurring

```python
img_gray=cv2.imread('images/dog.jpg',0)
```

```python
median_img=cv2.medianBlur(img_gray,15) #15->kernel size
```

```python
plt.figure(figsize=(16,9))
plt.subplot(121)
plt.imshow(img_gray,cmap='gray')
plt.subplot(122)
plt.imshow(median_img,cmap='gray')
```

```python
new_img_gray=img_gray[100:250,100:250]
```

```python
plt.imshow(new_img_gray,cmap='gray')
```

```python
kernel_size = 3
kernel = cv2.getGaussianKernel(kernel_size, 1.23)
gaussian_blurred_image = cv2.filter2D(new_img_gray, -1, kernel)
cv2.imshow('Gaussian Blurred image',gaussian_blurred_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

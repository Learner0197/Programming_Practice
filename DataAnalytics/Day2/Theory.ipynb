```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, KFold, cross_val_score, cross_val_predict, ShuffleSplit
from sklearn.utils import resample
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
import seaborn as sns

df=pd.read_csv('titanic.csv')
df=df[['survived','pclass','gender','age','sibsp','fare']].dropna()

le=LabelEncoder()
df['gender']=le.fit_transform(df['gender'])

x=df.drop('survived',axis=1)
y=df['survived']

#Bootstrapping
print('\n---------------Bootstrapping Example--------------')
n_iterations=100
n_size=int(len(x) * 0.8)
bootstrap_scores=[]

for i in range(n_iterations):
    x_resample,y_resample=resample(x,y,n_samples=n_size,random_state=i)
    x_train,x_test,y_train,y_test=train_test_split(x_resample,y_resample,test_size=0.2,random_state=i)
    model=LogisticRegression(max_iter=500)
    model.fit(x_train,y_train)
    y_pred=model.predict(x_test)
    acc=accuracy_score(y_test,y_pred)
    bootstrap_scores.append(acc)
print(f'Average accuracy over {n_iterations} bootstrapped samples: {np.mean(bootstrap_scores):.3f}')

#K-Fold Validation
print('\n--------------K-Fold-Cross-Validation--------------')
kf=KFold(n_splits=5,shuffle=True,random_state=1)
#kf=ShuffleSplit(n_splits=5,test_size=0.4,random_state=1)
model=LogisticRegression(max_iter=500)

#Perform K-Fold CV
cv_scores=cross_val_score(model,x,y,cv=kf,scoring='accuracy')

print(f'Accuracy scores for each fold: {cv_scores}')
print(f'Mean CV accuracy: {np.mean(cv_scores):.3f}')
```

```python
df = pd.read_csv("titanic.csv")

#Select relevant columns
df = df[['gender', 'survived']]
df.dropna(inplace=True)

#Total number of rows
total = len(df)
print(f"Totl Passengers: {total}")

#Marginal probability of gender
p_female = len(df[df['gender'] == 'female'])/total
p_male = len(df[df['gender'] == 'male'])/total

#Marginal probability of survival
p_survived = len(df[df['survived'] == 1])/total
p_not_survived = len(df[df['survived'] == 0])/total

print("\nMarginal Probabilities:")
print(f"P(gender = female): {p_female:.3f}")
print(f"P(gender = male): {p_male:.3f}")
print(f"P(survived = 1): {p_survived:.3f}")
print(f"P(survived = 0): {p_not_survived:.3f}")
```

```python
#Joint Probabilities
p_female_survived = len(df[(df['gender'] == 'female') & (df['survived'] == 1)])/total
p_male_survived = len(df[(df['gender'] == 'male') & (df['survived'] == 1)])/total
p_female_not_survived = len(df[(df['gender'] == 'female') & (df['survived'] == 0)])/total
p_male_not_survived = len(df[(df['gender'] == 'male') & (df['survived'] == 0)])/total
print("\nJoint Probabilities:")
print(f"P(gender = female, survived): {p_female_survived:.3f}")
print(f"P(gender = male, survived): {p_male_survived:.3f}")
print(f"P(gender = female, not survived): {p_female_not_survived:.3f}")
print(f"P(gender = male, not survived): {p_male_not_survived:.3f}")
```

```python
print("\nConditional Probabilities:")
#P(Survived | Female) = P(Survive and Female) / P(Female)
p_survived_given_female = p_female_survived / p_female
print(f"P(survived=1 | gender=female): {p_survived_given_female:.3f}")
p_survived_given_male = p_male_survived / p_male
print(f"P(survived=1 | gender=male): {p_survived_given_male:.3f}")
```

```python
#p_female_survived: This is a joint probability, which represents the probability the a randomly chosen passenger was both female and survived
#Calculated as: len(df[(df['gender'] == 'female) & (df['survived'] == 1)])/total
#p_survived_given_female: This is a conditional probability, which represents the probability that a randomly chosen passenger survived given that they were female
#Calculated as: p_female_survived / p_female
```

```python
#Use dataset https://kaggle.com/code/melikedilekci/diabetes-dataset-for-beginners/input
#Find the conditional probability that a person's Glucose levels are > 110 given that the person's BMI >= 25

df = pd.read_csv("diabetes.csv")
df = df[['Glucose', 'BMI']]
df.dropna(inplace=True)
total = len(df)
print(f"Total records: {total}")
p_glucose_above_110 = len(df[df['Glucose'] > 110])/total
p_bmi_above_25 = len(df[df['BMI'] >= 25])/total
p_glucose_above_110_and_bmi_above_25 = len(df[(df['Glucose'] > 110) & (df['BMI'] >= 25)])/total
print("\nMarginal Probabilities:")
print(f"P(Glucose > 110): {p_glucose_above_110:.3f}")
print(f"P(BMI >= 25): {p_bmi_above_25:.3f}")
print("\nJoint Probability:")
print(f"P(Glucose > 110 and BMI >= 25): {p_glucose_above_110_and_bmi_above_25:.3f}")
p_glucose_above_110_given_bmi_above_25 = p_glucose_above_110_and_bmi_above_25 / p_bmi_above_25
print("\nConditional Probability:")
print(f"P(Glucose > 110 | BMI >= 25): {p_glucose_above_110_given_bmi_above_25:.3f}")
```

```python
'''Different examples of binomial distribution:
Number of trials: n, Number of successes: k, Probability of success in each trial: p'''

'''Hospital records show that out of the patients admitted for specific disease, 75% die.
What is the probability that out of 6 patients admitted for the disease, 4 will recover?'''

'''Here n=6, k=4, p=100% - 75% = 25% = 0.25 (since 75% die, 25% recover)'''
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
probab=binom.pmf(k=4,n=6,p=0.25)
print('Probability that of six randomly selected patients, four will recover:',probab)
```

```python
'''A blindfolded marksman finds that on average, he hits the target 4 times out of 5. If he fires 4 shots, what is the probability that a) he will hit the target exactly 2 times? and b)at least 3 misses?'''
prob1=0
for k in range(3,5):
    prob1=prob1+binom.pmf(k=k,n=4,p=0.8)
print('Chance of hitting the target more than 2 times:',prob1)

prob2=0
for k in range(3,5):
    prob2=prob2+binom.pmf(k=k,n=4,p=0.2)
print('Chance of missing the target atleast 3 times:',prob2)

print('Chances calculated using cdf',binom.cdf(k=1,n=4,p=0.8))
```

```python
'''Normally 65% of all students who appear for C-DAC entrance test clear it. 50 students from a coaching class appeared for the C-DAC March 2024 entrance test. What is the probability that none of them will clear the test? What is the probability that more than 40 will clear the test?'''
'''n=50,k=var,p=0.65'''
prob_none=binom.cdf(k=0,n=50,p=0.65)
print('Probability that none of them will clear the test:',prob_none)

prob_more_than_40=1-binom.cdf(k=40,n=50,p=0.65)
print('Probability that more than 40 will clear the test:',prob_more_than_40)

# prob_method_2=binom.cdf(k=10,n=50,p=0.35)
# print('Probability that method 2 will clear the test:',prob_method_2)

prob_method_3=0
for k in range(41,51):
    prob_method_3=prob_method_3+binom.pmf(k=k,n=50,p=0.65)
print('Probability that method 3 will clear the test:',prob_method_3)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import zscore
from scipy import stats
df=pd.read_csv('diabetes.csv')
glucose_mean=df['Glucose'].mean()
glucose_std=df['Glucose'].std()
print(glucose_mean,glucose_std)

df['Glucose_zscore']=zscore(df['Glucose'])
print(df['Glucose_zscore'])

z_score_30=stats.norm.ppf(0.30)
print(z_score_30)
glucose_value_30=glucose_mean+z_score_30*glucose_std
print(f'Glucose value corresponding to 30th percentile: {glucose_value_30}')

z_score_given=2.8
glucose_value_given=glucose_mean+z_score_given*glucose_std
print(f'Glucose value for z-score of 2.8: {z_score_given}')

percentile_given=stats.norm.cdf(z_score_given)*100
print(f'Percentile given for z-score of 2.8: {percentile_given} %')
```

```python
#Check if data is normally ditributed shapiro test and qq plot
import matplotlib.pyplot as plt
import scipy.stats as stats
import numpy as np
import pandas as pd

data=pd.read_csv('tips.csv')
total_bill_series=data['total_bill']

#Normality test with Shapiro-Wilk test
shapiro_test=stats.shapiro(total_bill_series)
print('Test statistics (W):',shapiro_test.statistic)
print('p-value:',shapiro_test.pvalue)

alpha=0.05 #Common significance level
if shapiro_test.pvalue > alpha:
    print('Data is normally distributed')
else:
    print('Data is not normally distributed')

#QQ plot
stats.probplot(total_bill_series,dist='norm',plot=plt)
plt.title('QQ plot of total bill')
plt.xlabel('Theoretical quantiles(Normal Distribution)')
plt.ylabel('Ordered Values(Total Bill)')
plt.grid(True)
plt.show()
```

```python
from pulp import *

model = LpProblem("FactoryProductionProblem", LpMaximize)
chairs = LpVariable("Chairs", 0, None, LpInteger)
tables = LpVariable("Tables", 0, None, LpInteger)

model += 300 * chairs + 1000 * tables, "Profit"

model += 5 * chairs + 10 * tables <= 400, "Wood Constraint"
model += 2 * chairs + 6 * tables <= 300, "Time Constraint"
model += chairs >= 2, "Chairs Constraint"
model.solve()

for v in model.variables():
    print(v.name, "=", v.varValue)

print("Total Profit:", value(model.objective))
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

print(sns.get_dataset_names())

df = sns.load_dataset('flights')
print(df.head()) #We only get year and month for the date. We also need Day

#Convert the date to a YYYY-MM-DD format in a new column names yearMonth
df['yearMonth'] = "01-"+df['month'].astype(str)+"-"+df['year'].astype(str)
print(df.info())

#yearMonth is of type object - we may have problems later, so convert it into datetime
df['yearMonth'] = pd.to_datetime(df['yearMonth'])
print(df.info())
print(df.head())

#Make yearMonth column as the dataframe index
df.set_index('yearMonth', inplace=True)
print(df.head())

plt.figure(figsize = (10,5))
sns.lineplot(data=df, x=df.index, y=df.passengers)
plt.show()

from statsmodels.tsa.seasonal import seasonal_decompose
decomposition = seasonal_decompose(df.passengers, period=12)
fig = decomposition.plot()
plt.show()
```

```python
# df['rollMean'] = df.passengers.rolling(window=12).mean()
# df['rollStd'] = df.passengers.rolling(window=12).std()

def test_stationarity(dataFrame, var):
    dataFrame['rollMean'] = dataFrame[var].rolling(window=12).mean()
    dataFrame['rollStd'] = dataFrame[var].rolling(window=12).std()

    from statsmodels.tsa.stattools import adfuller
    adf_result = adfuller(dataFrame[var])
    p_value = adf_result[1]

    print(f"ADF p-value: {p_value:.4f}")
    if p_value < 0.05:
        print("The time series is stationary (reject H0).")
    else:
        print("The time series is not stationary (fail to reject H0).")

    plt.figure(figsize=(10,5))
    sns.lineplot(data=dataFrame, x=dataFrame.index, y=var, label='Original')
    sns.lineplot(data=dataFrame, x=dataFrame.index, y='rollMean', label='Rolling Mean')
    sns.lineplot(data=dataFrame, x=dataFrame.index, y='rollStd', label='Rolling Std')
    plt.title("Stationary check using Rolling Mean and Rolling standard deviation")
    plt.legend()
    plt.show()

test_stationarity(df,'passengers')
```

```python
#Monte Carlo Simulation

#Part 1:
def roll_dice():
    #Simulation of rolling a dice twice, minimum number we get is 1, max is 6
    #Add the results of the two simulations, so possible values will be:
    #(1,1) or (1,2) ... (6,6) ... Sum will be between 2 and 12
    #Run multiple times to verify
    return np.sum(np.random.randint(1,7,2)) #Generate an array of 2 random integers between 1 and 6 (both inclusive) and add them
print(roll_dice())

#Part 2:
#Someone approaches is saying I will give you 5 dollars if you get 7 and take 1 dollar if you get a number other than 7
#How do we know what will happen?
#Our own "Monte Carlo Simulation" like function
def monte_carlo_simulation(runs=1000):
    results = np.zeros(2) #An array, results[0] and results[1] initialised to two zeros
    for _ in range(runs): #Run 1000 times
        if roll_dice() == 7:
            results[0] += 1
        else:
            results[1] += 1
        return results
#Test 2-3 times and calculate how much you will win versus lose
print(monte_carlo_simulation())
print(monte_carlo_simulation())
print(monte_carlo_simulation())

#Part 3:
#Now do it 1000 times
#Takes some time

results = np.zeros(1000)

for i in range(1000):
    results[i] = monte_carlo_simulation()[0]

print(results)

#Ous win/loss:
print(results.mean())  #General Mean
print(results.mean()*5)  #What we will get as win on an average
```

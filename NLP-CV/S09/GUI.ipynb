```python
top=Tk()
top.title('My first application')
top.geometry('400x200')

def find():
    num=int(e.get())
    print('Square is:',num*num)
l=Label(top,text='Enter the number')
e=Entry(top)
b=Button(top,text='Find Square',command=find)

l.pack()
e.pack()
b.pack()
top.mainloop()
```

```python
top=Tk()
top.title('My first application')
top.geometry('400x200')

def find():
    num=int(e.get())
    print('Square is:',num*num)
    result=Label(top,text='Square is:'+str(num*num))
    result.pack()
l=Label(top,text='Enter the number')
e=Entry(top)
b=Button(top,text='Find Square',command=find)

l.pack()
e.pack()
b.pack()
top.mainloop()
```

```python
top=Tk()
top.title('My first application')
top.geometry('400x200')

def find():
    num=int(e.get())
    result.config(text='Square is:'+str(num*num))
l=Label(top,text='Enter the number')
e=Entry(top)
b=Button(top,text='Find Square',command=find)
result=Label(top,text='Square is:')

l.pack()
e.pack()
b.pack()
result.pack()
top.mainloop()
```

```python
top=Tk()
top.title('My first application')
top.geometry('400x250')

def find():
    num=int(e.get())
    result.config(text='Square is:'+str(num*num))
l=Label(top,text='Enter the number')
e=Entry(top)
b=Button(top,text='Find Square',command=find)
result=Label(top,text='Square is:')

l.place(x=50,y=50)
e.place(x=50,y=50)
b.place(x=50,y=110)
result.place(x=50,y=170)
top.mainloop()
```

```python
top=Tk()
top.title('Arithmetic operations')
top.geometry('600x500')

def add():
    num1=int(e1.get())
    num2=int(e2.get())
    result.config(text='Sum is:'+str(num1+num2))
def sub():
    num1=int(e1.get())
    num2=int(e2.get())
    result.config(text='Difference is:'+str(num1-num2))
def mul():
    num1=int(e1.get())
    num2=int(e2.get())
    result.config(text='Product is:'+str(num1*num2))
def div():
    num1=int(e1.get())
    num2=int(e2.get())
    result.config(text='Quotient is:'+str(num1/num2))
l1=Label(top,text='Enter 1st number')
l2=Label(top,text='Enter 2nd number')
e1=Entry(top)
e2=Entry(top)
b1=Button(top,text='Add',command=add)
b2=Button(top,text='Subtract',command=sub)
b3=Button(top,text='Multiply',command=mul)
b4=Button(top,text='Divide',command=div)
result=Label(top,text='Answer is:')

l1.place(x=50,y=50)
l2.place(x=50,y=70)
e1.place(x=150,y=50)
e2.place(x=150,y=70)
b1.place(x=50,y=170)
b2.place(x=150,y=170)
b3.place(x=250,y=170)
b4.place(x=350,y=170)
result.place(x=50,y=250)
top.mainloop()
```

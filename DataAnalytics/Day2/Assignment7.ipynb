## 1. Write a Pandas program to create
a) Datetime object for Jan 15 2012.
b) Specific date and time of 9:20 pm.
c) Local date and time.
d) A date without time.
e) Current date.
f) Time from a datetime.
g) Current local time.

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
date_a = pd.Timestamp('2012-01-15')
print(f"a) {date_a}")
date_b = pd.Timestamp('2012-01-15 21:20:00')
print(f"b) {date_b}")
date_c = pd.Timestamp.now()
print(f"c) {date_c}")
date_d = pd.Timestamp('2012-01-15').date()
print(f"d) {date_d}")
date_e = pd.Timestamp.today().date()
print(f"e) {date_e}")
date_f = pd.Timestamp('2012-01-15 21:20:00').time()
print(f"f) {date_f}")
date_g = pd.Timestamp.now().time()
print(f"g) {date_g}")
```

```python
date_a=pd.Timestamp(year=2023,month=11,day=19)
print(f"Date: {date_a}")
day_after=date_a + pd.Timedelta(days=1)
print(f"Day after: {day_after}")
day_before=date_a - pd.Timedelta(days=1)
print(f"Day before: {day_before}")
date_b=pd.Timestamp(year=2023,month=11,day=25)
days_between=(date_b - date_a).days
print(f"Days between two dates: {days_between}")
```

# 17. Write a Pandas program to convert unix/epoch time to a regular time stamp in UTC. Also convert the said timestamp in to a given time zone.

```python
epoch_time = 1589500800
utc_time = pd.to_datetime(epoch_time, unit='s')
print(f"UTC time: {utc_time}")
local_time = utc_time.tz_localize('UTC').tz_convert('US/Eastern')
print(f"Eastern time: {local_time}")
```

# 27. Write a Pandas program to calculate one, two, three business day(s) from a specified date. Also find the next business month end from a specific date.

```python
date = pd.Timestamp('2020-05-15')
one_business_day = date + pd.offsets.BusinessDay(1)
two_business_days = date + pd.offsets.BusinessDay(2)
three_business_days = date + pd.offsets.BusinessDay(3)
print(f"1 business day: {one_business_day}")
print(f"2 business days: {two_business_days}")
print(f"3 business days: {three_business_days}")
next_month_end = date + pd.offsets.BMonthEnd()
print(f"Next business month end: {next_month_end}")
```

# 29. Write a Pandas program create a series with a PeriodIndex which represents all the calendar month periods in 2029 and 2031. Also print the values for all periods in 2030.
Note: PeriodIndex is an immutable ndarray holding ordinal values indicating regular periods in time such as particular years, quarters, months, etc.

```python
periods = pd.period_range(start='2029-01', end='2031-12', freq='M')
print(f"All periods:\n{periods}")
periods_2030 = periods[periods.year == 2030]
print(f"\nPeriods in 2030:\n{periods_2030}")
```

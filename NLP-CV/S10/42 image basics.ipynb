```python
img_chess = np.zeros([8,8], dtype=np.uint8)
```

```python
for count in range(8):
    for x in range(8):
        if (count+x) % 2 == 0:
            img_chess[count, x] = 1
```

```python
import matplotlib.pyplot as plt
img_plt = plt.imread('images/dog.jpg')

img_new = np.array(img_plt, copy=True)

for i in range(img_new.shape[0]):
    for j in range(img_new.shape[1]):
        if j == i or j + i == img_new.shape[0] - 1:
            img_new[i, j] = [0, 0, 0]

plt.imshow(img_new)
```

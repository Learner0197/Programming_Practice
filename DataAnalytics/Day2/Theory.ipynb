```python
import pandas as pd
from numpy.conftest import dtype

# Boolean Indexing
dataset=pd.read_csv('tips.csv')
filtered1=dataset[(dataset['gender']=='Female') & (dataset['tip']>=4)]
filtered1
```

```python
#Using Query
filtered2=dataset.query('gender=="Female" and tip>=4')
filtered2
```

```python
#Using loc with conditions
filtered3=dataset.loc[(dataset['gender']=='Female') & (dataset['tip']>=4)]
filtered3
```

```python
#Using where(return NaN where condition is False)
filtered4=dataset.where((dataset['gender']=='Female') & (dataset['tip']>=4))
filtered4=filtered4.dropna()
filtered4
```

```python
df=dataset.copy()
df_grouped=df.groupby('gender')
print(df_grouped)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('MS_Dhoni_ODI_record.csv')
print(df.head())
print(df.tail())
```

```python
df['opposition']=df['opposition'].apply(lambda x: x[2:])
#df['opposition']=df['opposition'].str.replace('v ','',regex=False)
print(df['opposition'].head())
```

```python
df['date']=pd.to_datetime(df['date'],dayfirst=True)#dayfirst used to change date format in the date section
df['year']=df['date'].dt.year.astype(int)
print(df.head())
```

```python
df['score']=df['score'].apply(str)
df['not out']=np.where(df['score'].str.endswith('*'),1,0)#check not out using * and assign with 1 or 0
print(df.head())
```

```python
df.drop(columns='odi_number',inplace=True)#makes change to original dataframe using inplace
print(df.head())
```

```python
df_new=df.loc[((df['score']!='DNB') & (df['score']!='TDNB')),'runs_scored':]
print(df_new.head())
```

```python
df_new['runs_scored']=df_new['runs_scored'].astype(int)
df_new['balls_faced']=df_new['balls_faced'].astype(int)
df_new['strike_rate']=df_new['strike_rate'].astype(float)
df['fours']=df_new['fours'].astype(int)
df_new['sixes']=df_new['sixes'].astype(int)
print(df_new.head())
```

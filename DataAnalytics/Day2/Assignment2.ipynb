## Question 10: Write a Pandas program to convert Series of lists to one Series.

```python
ser = pd.Series([['Red', 'Green', 'White'], ['Red', 'Black'], 'Yellow'])
ser2 = pd.Series()
for i in ser:
    if isinstance(i, list):
        for j in i:
            ser2 = pd.concat([ser2, pd.Series([j])], ignore_index=True)
    else:
        ser2 = pd.concat([ser2, pd.Series([i])], ignore_index=True)
ser2
```

## Question 11: Write a Pandas program to sort a given Series.

```python
s = pd.Series([100, 200, 'python', 300.12, 400])
l = []
ls = []
for i in s:
    if type(i) != str:
        l.append(i)
    else:
        ls.append(i)

ln = sorted(l) + sorted(ls)
s_sorted = pd.Series(ln)
s_sorted
```

## Question 28: Write a Pandas program to get the day of month, day of year, week number and day of week from a given series of date strings.

## Question 30: Write a Pandas program to filter words from a given series that contain atleast two vowels.

```python
s = pd.Series(['Red', 'Green', 'Orange', 'Pink', 'Yellow', 'White'])
vowels = 'aeiouAEIOU'
result = s[s.apply(lambda x: sum(c in vowels for c in x) >= 2)]
print(s)
print(result)
```

## Question 31: Write a Pandas program to compute the Euclidean distance between two given series.

```python
s1 = pd.Series(range(1, 11))
s2 = pd.Series([11, 8, 7, 5, 6, 5, 3, 4, 7, 1])
distance = np.sqrt(((s1 - s2) ** 2).sum())
print(s1)
print(s2)
print(distance)
```

## Question 32: Write a Pandas program to find the positions of the values neighboured by smaller values on both sides in a given series.

```python
s = pd.Series([1, 8, 7, 5, 6, 5, 3, 4, 7, 1])
positions = []
for i in range(1, len(s) - 1):
    if s[i] > s[i-1] and s[i] > s[i+1]:
        positions.append(i)
print(s)
print(np.array(positions))
```

## Question 33: Write a Pandas program to replace missing white spaces in a given string with the least frequent character.

```python
s = 'abc def abcdef icd'
freq = pd.Series(list(s)).value_counts()
least_freq = freq.idxmin()
print(freq)
result = s.replace(' ', least_freq)
print(result)
```

## Question 34: Write a Pandas program to compute the autocorrelations of a given numeric series.

```python
s = pd.Series(np.random.randn(15))
autocorr = [s.autocorr(lag=i) for i in range(1, 11)]
autocorr = [round(x, 2) for x in autocorr]
print(s)
print(autocorr)
```

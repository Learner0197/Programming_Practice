```python
import nltk
import string
from nltk import word_tokenize, sent_tokenize
```

```python
sent='Hello friends! How are you? Welcome to "Python" Programming.'
```

```python
sent=word_tokenize(sent)
sent
```

```python
for i in sent:
    if not i.isalnum():
        sent.remove(i)
```

```python
sent
```

```python
clean=' '.join(sent)
clean
```

```python
new_sent=''
for char in sent:
    if char.isalpha() or char.isdigit() or char.isspace():
        new_sent += char
new_sent
```

Stopwords Removal

```python
from nltk.corpus import stopwords
swords=stopwords.words('english')
```

```python
swords
```

```python
new_sent=new_sent.lower().split()
for words in new_sent:
    if words in swords:
        new_sent.remove(words)
```

```python
new_sent=clean
```

```python
clean_text=[word for word in word_tokenize(new_sent) if word.lower() not in swords]
" ".join(clean_text)
```

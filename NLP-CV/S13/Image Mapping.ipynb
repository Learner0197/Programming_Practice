```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img=cv2.imread(r'C:\Users\dai\PycharmProjects\NLP-CV\S12\images\dog.jpg')
```

```python
rows,cols=img.shape[:2]
```

```python
rows,cols
```

```python
#create a grid
gridx,gridy=np.meshgrid(np.arange(cols),np.arange(rows))
```

```python
gridy
```

```python
mapx=gridx
mapy=gridy+25*np.sin(gridx/20)
```

```python
plt.plot(np.arange(1,40,0.1),np.sin(np.arange(1,40,0.1)))
```

```python
mapx=mapx.astype(np.float32)
mapy=mapy.astype(np.float32)
```

```python
mapped_img=cv2.remap(img,mapx,mapy,cv2.INTER_LINEAR)
```

```python
mapped_image=cv2.cvtColor(mapped_img,cv2.COLOR_BGR2RGB)
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

```python
plt.subplot(1,2,1)
plt.imshow(img)
plt.subplot(1,2,2)
plt.imshow(mapped_image)
```

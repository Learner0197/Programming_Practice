## Q.1
## sent = 'They told that their ages are 23, 25 and 29 respectively.'
### 1. Find the average of ages.
### 2. Count total number of words starts with 't' / 'T'
### 3. Count total punctuation symbols.
### 4. Print the words in capital letter which ends with 'd'
### 5. Merge all the words and print (no punctuations)

```python
from string import punctuation

from nltk import word_tokenize
```

```python
sent = 'They told that their ages are 23, 25 and 29 respectively.'
```

```python
num = [int(i) for i in word_tokenize(sent) if i.isnumeric()]
print('Average of numbers = ', sum(num)/len(num))
```

```python
count = [s for s in word_tokenize(sent.lower()) if s.startswith('t')]
print('Count of words starting with t/T = ', len(count))
```

```python
import string
count = [p for p in word_tokenize(sent) if p in string.punctuation]
print('Count of punctuation = ', len(count))
```

```python
capital = [s.upper() for s in word_tokenize(sent) if s.endswith('d')]
print("Words ending with 'd' in capital: ", capital)
```

```python
no_punc = [p for p in word_tokenize(sent) if p not in string.punctuation]
print('all words merged without punctuations:', ''.join(no_punc))
```

### Open the wikipedia page of Indian flag and perform following analysis
### Count total number of Words
### Count percentage of punctuation marks
### Print total number of lines
### How many words are in title case?
### Remove the numbers and print the text

```python
india='''The national flag of India, colloquially called the Tiraṅgā (tricolour), is a horizontal rectangular tricolour flag, the colours being of India saffron, white and India green; with the Ashoka Chakra, a 24-spoke wheel, in navy blue at its centre.[1][2] It was adopted in its present form during a meeting of the Constituent Assembly held on 22 July 1947, and it became the official flag of the Union of India on 15 August 1947. The flag was subsequently retained as that of the Republic of India. In India, the term "tricolour" almost always refers to the Indian national flag.
The current Indian flag was designed by Pingali Venkayya based on the Swaraj flag, a flag of the Indian National Congress adopted by Mahatma Gandhi after making significant modifications to the design proposed by Pingali Venkayya.[3] This flag included the charkha which was replaced with the chakra in 1947 by Tyabji.[4]
Before the amendment of the flag code in 2021, the flag was by law only to be made of khadi; a special type of hand-spun cloth or silk, made popular by Mahatma Gandhi. The manufacturing process and specifications for the flag are laid out by the Bureau of Indian Standards. The right to manufacture the flag is held by the Khadi Development and Village Industries Commission, which allocates it to regional groups. As of 2023, there are four units in India that are licensed to manufacture the flag.
Usage of the flag is governed by the Flag Code of India and other laws relating to the national emblems. The original code prohibited use of the flag by private citizens except on national days such as the Independence day and the Republic Day. In 2002, on hearing an appeal from a private citizen, Naveen Jindal, the Supreme Court of India directed the Government of India to amend the code to allow flag usage by private citizens. Subsequently, the Union Cabinet of India amended the code to allow limited usage. The code was amended once more in 2005 to allow some additional use including adaptations on certain forms of clothing. The flag code also governs the protocol of flying the flag and its use in conjunction with other national and non-national flags.'''
```

```python
tokenized=word_tokenize(india)
```

```python
tokenized
```

```python
print(len(tokenized))
```

```python
punctuations=0
for words in tokenized:
    if not words.isalnum():
        punctuations=punctuations+1
punctuation_percentage=punctuations/len(india)*100
print(punctuation_percentage)
```

```python
line_count=india
total_lines=line_count.strip().count('\n') + 1
print(total_lines)
```

```python
words=word_tokenize(india)
title_case_words=[word for word in words if word.istitle()]
print(len(title_case_words))
```

```python
without_numbers=[word for word in words if not word.isdigit()]
text_without_numbers=" ".join(without_numbers)
print(text_without_numbers)
```

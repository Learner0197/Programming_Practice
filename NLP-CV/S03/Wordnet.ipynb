```python
#Extract specific POS from the word
verbs = set()
nouns = set()
adjts = set()

for word in wordnet.synsets('sound', pos='v'):
    verbs.add(word.lemmas()[0].name())
for word in wordnet.synsets('sound', pos='n'):
    nouns.add(word.lemmas()[0].name())
for word in wordnet.synsets('sound', pos='a'):
    adjts.add(word.lemmas()[0].name())
print("Verb synonyms:")
print(verbs)
print("Noun synonyms:")
print(nouns)
print("Adject synonyms:")
print(adjts)
```

```python
def ext_syn(word, pos, cat):
    for w in wordnet.synsets(word, pos=pos):
        cat.add(w.lemmas()[0].name())
    return cat

ext_syn('sound', 'v', verbs)
```

```python
ext_syn('sound', 'n', nouns)
```

```python
ext_syn('sound', 'a', adjts)
```

```python
# Retrieve the antonyms = set()

antonyms = set()
for word in wordnet.synsets('possible'):
    for lemma in word.lemmas():
        if lemma.antonyms():
            antonyms.add(lemma.antonyms()[0].name())
```

### Extract the hypernyms

```python
word='soil'
synsets=wordnet.synsets(word)

for x in synsets:
    hypernyms = x.hypernyms()
    for h in hypernyms:
        print(word, 'hypernym of', h.name())
```

```python
word='intelligence'
synsets=wordnet.synsets(word)

for x in synsets:
    hypernyms = x.hypernyms()
    for h in hypernyms:
        print(word, 'hypernym of', h.name())
```

```python
# Extract Meronyms

word = 'car'
synsets=wordnet.synsets(word)
for x in synsets:
    meronyms = x.part_meronyms()
    for m in meronyms:
        print(word, 'meronym of', m.name())
```

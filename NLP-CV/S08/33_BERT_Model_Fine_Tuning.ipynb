```python
import pandas as pd
import tensorflow as tf
import numpy as np
import ktrain
from ktrain import text
```

```python
df = pd.read_csv('SMSSpamCollection', sep='\t', names = ['labels', 'text'])
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
data_train, data_test = train_test_split(df, test_size=0.2, random_state=0)
```

```python
data_train.shape, data_test.shape
```

```python
(x_train, y_train), (x_test, y_test), preproc = text.texts_from_df(train_df=data_train, text_column = 'text', label_columns = 'labels',
                                                                   val_df = data_test,
                                                                   preprocess_mode = 'bert')
```

```python
y_train
```

```python
model = text.text_classifier(name = 'bert', train_data=(x_train, y_train), preproc=preproc)
```

```python
type(model)
```

```python
learner = ktrain.get_learner(model=model, train_data=(x_train, y_train), val_data=(x_test, y_test), batch_size=6)
```

```python
learner
```

```python
# learner.lr_find(show_plot=True, max_epochs=2)
```

```python
learner.fit_onecycle(lr = 2e-5, epochs=1)
```

```python
predictor = ktrain.get_predictor(learner.model, preproc)
```

```python
sent = 'You won bonus prize of 10000. Call us to claim.'
```

```python
predictor.predict(sent)
```

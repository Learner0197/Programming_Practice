## 1. Write a Python programming to display a bar chart of the popularity of programming Languages.
Sample data:
Programming languages: Java, Python, PHP, JavaScript, C#, C++
Popularity: 22.2, 17.6, 8.8, 8, 7.7, 6.7

```python
lang=['Java','Python','C','C++','JavaScript','Php']
popularity=[22.2, 17.6, 8.8, 8, 7.7, 6.7]
plt.bar(lang,popularity,align='center',color='r')
plt.xticks(lang,lang)
plt.title('Bar Chart')
plt.xlabel('Language')
plt.ylabel('Popularity')
plt.grid(True,which='major',color='r',linestyle='--')
plt.grid(True,which='minor',color='b',linestyle='-')
plt.minorticks_on()

plt.show()
```

## 10. Write a Python program to create bar plot of scores by group and gender. Use multiple X values on the same chart for men and women.
Sample Data:
Means (men) = (22, 30, 35, 35, 26)
Means (women) = (25, 32, 30, 35, 29)

```python
import matplotlib.pyplot as plt
import numpy as np
men_means = (22, 30, 35, 35, 26)
women_means = (25, 32, 30, 35, 29)
labels = ['Group 1', 'Group 2', 'Group 3', 'Group 4', 'Group 5']
n_groups = len(labels)
index = np.arange(n_groups)
bar_width = 0.35
fig, ax = plt.subplots(figsize=(10, 6))
rects1 = ax.bar(index - bar_width/2, men_means, bar_width, label='Men', color='skyblue')
rects2 = ax.bar(index + bar_width/2, women_means, bar_width, label='Women', color='lightcoral')
ax.set_xlabel('Group')
ax.set_ylabel('Scores')
ax.set_title('Scores by Group and Gender')
ax.set_xticks(index)
ax.set_xticklabels(labels)
ax.legend()
plt.tight_layout()
plt.show()
```

## 14. Write a Python program to create a stacked bar plot with error bars. Note: Use bottom to stack the women?s bars on top of the men?s bars.
Sample Data:
Means (men) = (22, 30, 35, 35, 26)
Means (women) = (25, 32, 30, 35, 29)
Men Standard deviation = (4, 3, 4, 1, 5)
Women Standard deviation = (3, 5, 2, 3, 3)

```python
men_means = (22, 30, 35, 35, 26)
women_means = (25, 32, 30, 35, 29)
men_std = (4, 3, 4, 1, 5)
women_std = (3, 5, 2, 3, 3)
categories = ['G1', 'G2', 'G3', 'G4', 'G5']
x = np.arange(len(categories))
plt.bar(x, men_means, yerr=men_std, capsize=5, label='Men', color='skyblue')
plt.bar(x, women_means, yerr=women_std, capsize=5, bottom=men_means, label='Women', color='pink')
plt.xlabel('Groups')
plt.ylabel('Scores')
plt.title('Stacked Bar Plot with Error Bars')
plt.xticks(x, categories)
plt.legend()
plt.show()
```

## 15. Write a Python program to create a horizontal bar chart with differently ordered colors. Note: Use bottom to stack the women?s bars on top of the men?s bars.
Sample Data Set:
languages = [['Language','Science','Math'],
['Science','Math','Language'],
['Math','Language','Science']]
numbers = [{'Language':75, 'Science':88, 'Math':96},
{'Language':71, 'Science':95, 'Math':92},
{'Language':75, 'Science':90, 'Math':89}]

```python
languages = [['Language','Science','Math'],
             ['Science','Math','Language'],
             ['Math','Language','Science']]
numbers = [{'Language':75, 'Science':88, 'Math':96},
           {'Language':71, 'Science':95, 'Math':92},
           {'Language':75, 'Science':90, 'Math':89}]
categories = ['Student 1', 'Student 2', 'Student 3']
y = np.arange(len(categories))
for i in range(len(categories)):
    left = 0
    for subject in languages[i]:
        plt.barh(y[i], numbers[i][subject], left=left, label=subject if i == 0 else "")
        left += numbers[i][subject]
plt.ylabel('Students')
plt.xlabel('Scores')
plt.title('Horizontal Stacked Bar Chart')
plt.yticks(y, categories)
plt.legend(['Language', 'Science', 'Math'])
plt.show()
```

## 16. Write a Python program to create stack bar plot and add label to each section.
Sample data:
people = ('G1','G2','G3','G4','G5','G6','G7','G8')
segments = 4
## multi-dimensional data
data = [[ 3.40022085, 7.70632498, 6.4097905, 10.51648577, 7.5330039,
7.1123587, 12.77792868, 3.44773477],
[ 11.24811149, 5.03778215, 6.65808464, 12.32220677, 7.45964195,
6.79685302, 7.24578743, 3.69371847],
[ 3.94253354, 4.74763549, 11.73529246, 4.6465543, 12.9952182,
4.63832778, 11.16849999, 8.56883433],
[ 4.24409799, 12.71746612, 11.3772169, 9.00514257, 10.47084185,
10.97567589, 3.98287652, 8.80552122]]

```python
people = ('G1','G2','G3','G4','G5','G6','G7','G8')
segments = 4
data = [[3.40022085, 7.70632498, 6.4097905, 10.51648577, 7.5330039, 7.1123587, 12.77792868, 3.44773477],
        [11.24811149, 5.03778215, 6.65808464, 12.32220677, 7.45964195, 6.79685302, 7.24578743, 3.69371847],
        [3.94253354, 4.74763549, 11.73529246, 4.6465543, 12.9952182, 4.63832778, 11.16849999, 8.56883433],
        [4.24409799, 12.71746612, 11.3772169, 9.00514257, 10.47084185, 10.97567589, 3.98287652, 8.80552122]]
colors = ['red', 'green', 'blue', 'orange']
y = np.arange(len(people))
left = np.zeros(len(people))
for i in range(segments):
    # Use plt.barh() for a horizontal bar plot
    bars = plt.barh(y, data[i], left=left, color=colors[i], label=f'Segment {i+1}')
    for j, bar in enumerate(bars):
        width = bar.get_width()
        plt.text(left[j] + width/2, bar.get_y() + bar.get_height()/2,
                 f'{data[i][j]:.1f}', ha='center', va='center', fontsize=8)
    left += data[i]
plt.ylabel('Groups')
plt.xlabel('Values')
plt.title('Horizontal Stacked Bar Plot with Labels')
plt.yticks(y, people)
plt.legend()
plt.show()
```

## 17. Write a Python program to add textures (black and white) to bars and wedges. Note: Use bottom to stack the womens bars on top of the mens bars.

```python
fig = plt.figure()
patterns = [ "/" , "\\" , "|" , "-" , "+" , "x", "o", "O", ".", "*" ]
ax1 = fig.add_subplot(111)
for i in range(len(patterns)):
    ax1.bar(i, 3, color='red', edgecolor='black', hatch=patterns[i])
plt.show()
```

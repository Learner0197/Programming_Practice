```python
import string, re, nltk, os
from nltk.corpus import stopwords
from tensorflow.keras.preprocessing.text import Tokenizer
import numpy as np
from tf_keras.src.layers.preprocessing.benchmarks.feature_column_benchmark import create_vocabulary
```

```python
# Load the doc in memory
def load_doc(filename):
    f = open(filename)
    text = f.read()
    f.close()
    return text
```

```python
from nltk.tokenize import word_tokenize
swords = stopwords.words('english')
```

```python
# function to clean the doc
def clean_doc(doc):
    tokens = word_tokenize(doc)
    tokens = [token for token in tokens if token.isalpha()]
    tokens = [token for token in tokens if token not in swords]
    tokens = [token for token in tokens if len(token) > 1]
    return tokens
```

```python
#Load the data, clean the data and return lines
def doc_to_line(filename):
    doc = load_doc(filename)
    tokens = clean_doc(doc)
    tokens = [token for token in tokens if token in vocab]
    return " ".join(tokens)
```

```python
# load all the files from the directory
def process_train(directory):
    documents = []
    for filename in os.listdir(directory):
        if not filename.startswith('cv9'):
            path = directory + '//' + filename
            doc = load_doc(path)
            tokens = clean_doc(doc)
            documents.append(tokens)
    return documents
```

```python
def process_docs(directory, is_train):
    documents = []
    for filename in os.listdir(directory):
        if is_train and filename.startswith('cv9'):
            continue
        if not is_train and not filename.startswith('cv9'):
            continue
        path = directory + '//' + filename
        docs = load_doc(path)
        tokens = clean_doc(docs)
        documents.append(tokens)
    return documents
```

```python
def load_data(is_train):
    neg = process_docs(r'C:\Users\dai\PycharmProjects\NLP-CV\S05\review_polarity-20251128T094730Z-1-001\review_polarity\txt_sentoken\neg', is_train)
    pos = process_docs(r'C:\Users\dai\PycharmProjects\NLP-CV\S05\review_polarity-20251128T094730Z-1-001\review_polarity\txt_sentoken\pos', is_train)
    docs = neg + pos
    labels = [0 for i in range(len(neg))] + [1 for i in range(len(pos))]
    return docs, labels
```

```python
# load training data
train_data, train_labels = load_data(True)
# load testing data
test_data, test_labels = load_data(False)
```

### Data preparation

```python
def create_tokenizer(lines):
    tokenizer = Tokenizer()
    tokenizer.fit_on_texts(lines)
    return tokenizer
```

```python
tokenizer = create_tokenizer(train_data)
```

```python
# encode the training data
x_train = tokenizer.texts_to_matrix(train_data)
```

```python
x_train.shape
```

```python
len(tokenizer.word_index)
```

```python
# encode the testing data
x_test = tokenizer.texts_to_matrix(test_data)
```

```python
x_test.shape
```

### Build the model

```python
from keras.models import Sequential
from keras.layers import Dense, Input
from keras.utils import plot_model
```

```python
def define_model(n_words):
    model = Sequential()
    model.add(Input(shape=(n_words,)))
    model.add(Dense(100, activation='relu'))
    model.add(Dense(1, activation='sigmoid'))
    model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
    return model
```

```python
model = define_model(36389)
```

```python
plot_model(model, show_layer_activations=True, show_layer_names=True, show_shapes=True)
```

### Train the model

```python
model.fit(x_train, np.array(train_labels), epochs=10, batch_size=5)
```

```python
model.evaluate(x_test, np.array(test_labels), batch_size=1)
```

```python
text1 = 'Best movie ever! It was great, I will definitely recomment it.'
text2 = "This is the bad movie. Please don't watch it."
```

```python
def predict_sentiment(review):
    tokens=clean_doc(review)
    # line = " ".join(tokens)
    encoded=tokenizer.texts_to_matrix(tokens)
    yhat=model.predict(encoded)
    prob = yhat[0,0]
    if prob>0.5:
        return 'Positive'
    return 'Negative'
```

```python
predict_sentiment(text1)
```

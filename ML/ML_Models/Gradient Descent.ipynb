## Gradient Descent using Min-Max

```python
import numpy as np
import pandas as pd
from sklearn.metrics import log_loss,classification_report
from sklearn.preprocessing import MinMaxScaler,LabelEncoder
from sklearn.model_selection import train_test_split
kyph=pd.read_csv('Kyphosis.csv')
le=LabelEncoder()
kyph['Kyphosis']=le.fit_transform(kyph['Kyphosis'])
y=kyph['Kyphosis']
x=kyph.drop('Kyphosis',axis=1)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,stratify=y,random_state=25)
mm=MinMaxScaler()
x_trn_scl=mm.fit_transform(x_train)
x_tst_scl=mm.fit_transform(x_test)
x_trn_scl.shape
x_trn_scl=pd.DataFrame(x_trn_scl,columns=x_train.columns)
#x_tst_scl=pd.DataFrame(x_trn_scl,columns=x_test.columns)

x_trn_scl
```

```python
x1,x2,x3=x_trn_scl['Age'],x_trn_scl['Number'],x_trn_scl['Start']
eta=0.3
b,w1,w2,w3=0.1,0.1,0.1,0.1
epochs=1000
for i in range(epochs):
    ycap=sigmoid(w1*x1+w2*x2+w3*x3+b)
    L=log_loss(y_train,ycap)
    if i%100==0:
        print(f'W1: {w1}, W2: {w2}, W3: {w3}, B: {b}, Loss: {L}')
    dw1=-np.mean((y_train-ycap)*x1)
    dw2=-np.mean((y_train-ycap)*x2)
    dw3=-np.mean((y_train-ycap)*x3)
    db=-np.mean(y_train-ycap)
    w1=w1-eta*dw1
    w2=w2-eta*dw2
    w3=w3-eta*dw3
    b=b-eta*db
```

## Equivalent code to .pred_proba()

```python
y_pred_prob = sigmoid(w1*x_tst_scl[:, 0] + w2*x_tst_scl[:, 1] + w3*x_tst_scl[:, 2] + b)
print('Log loss on test set:', log_loss(y_test, y_pred_prob))
```

```python
import warnings; warnings.filterwarnings('ignore')
y_pred=np.where(y_pred_prob>0.5,1,0)
print(classification_report(y_test, y_pred))
```

```python
x=np.array([1.2,1.4,0.5,0.6,1.1,1.3])
y=np.array([1,1,0,0,1,1])
init_b,init_w=0.1,0.1
eta=0.1
```

## Batch Gradient Descent

```python
w,b=init_w,init_b
tol=0.01
for i in range(0,100000):
    yhat=sigmoid(x*w+b)
    L=log_loss(y,yhat)
    if L<tol:
        print(f'Algo stopped at {i+1}th iteration')
        print(f'w= {w}, b= {b}')
        print(f'Loss: {L}')
        break
    db=-np.mean(y-yhat)
    dw=-np.mean((y-yhat)*x)
    w=w-eta*dw
    b=b-eta*db
```

## Stochastic Gradient Descent

```python
w,b=init_w,init_b
tol=0.01
eta=0.1
for i in range(0,100000):
    for j in range(0,len(x)):
        yhat=sigmoid(x[j]*w+b)
        db=-(y[j]-yhat)
        dw=-((y[j]-yhat)*x[j])
        w=w-eta*dw
        b=b-eta*db

    yhat_all=sigmoid(x*w+b)
    L=log_loss(y,yhat_all)
    if L<tol:
        print(f'Algo stopped at {i+1}th iteration at Loss < Tolerance {tol}')
        print(f'w= {w}, b= {b}')
        print(f'Loss: {L}')
        break
```

## Min-Batch Gradient Descent

```python
batches=([0,1],[2,3],[4,5])
y[batches[1]]
```

```python
w,b=init_w,init_b
tol=0.01
eta=0.1
for i in range(0,100000):
    for batch in batches:
        yhat=sigmoid(x[batch]*w+b)
        db=-np.mean(y[batch]-yhat)
        dw=-np.mean((y[batch]-yhat)*x[batch])
        w=w-eta*dw
        b=b-eta*db

    yhat_all=sigmoid(x*w+b)
    L=log_loss(y,yhat_all)
    if L<tol:
        print(f'Algo stopped at {i+1}th iteration at Loss < Tolerance {tol}')
        print(f'w= {w}, b= {b}')
        print(f'Loss: {L}')
        break
```

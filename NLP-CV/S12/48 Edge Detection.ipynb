```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread('images/lenna.png',0)
```

```python
help(cv2.Sobel)
```

```python
sobelx=cv2.Sobel(img,cv2.CV_64F,1,0,ksize=3)
sobely=cv2.Sobel(img,cv2.CV_64F,0,1,ksize=3)
```

```python
sobelx.dtype
```

```python
final_sobel=cv2.add(sobelx,sobely)
```

```python
thresh,binary_sobel=cv2.threshold(final_sobel,100,255,cv2.THRESH_BINARY)
```

```python
plt.figure(figsize=[16,9])
plt.subplot(121)
plt.title('Original')
plt.imshow(img,cmap='gray')
plt.subplot(122)
plt.title('Edges')
plt.imshow(binary_sobel,cmap='gray')
```

```python
thresh,sobelx=cv2.threshold(sobelx,60,255,cv2.THRESH_BINARY)
thresh,sobely=cv2.threshold(sobely,60,255,cv2.THRESH_BINARY)
```

```python
plt.figure(figsize=[16,9])
plt.subplot(141)
plt.title('Original')
plt.imshow(img,cmap='gray')
plt.subplot(142)
plt.title('Sobel X')
plt.imshow(sobelx,cmap='gray')
plt.subplot(143)
plt.title('Sobel Y')
plt.imshow(sobely,cmap='gray')
plt.subplot(144)
plt.title('Final')
plt.imshow(binary_sobel,cmap='gray')
```

```python
img=cv2.imread('images/circle.png',0)
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
final_sobel = cv2.add(sobelx, sobely)
thresh, binary_sobel = cv2.threshold(final_sobel, 100, 255, cv2.THRESH_BINARY)
plt.figure(figsize=[16,9])
plt.subplot(121)
plt.title('Original')
plt.imshow(img,cmap='gray')
plt.subplot(122)
plt.title('Edges')
plt.imshow(binary_sobel,cmap='gray')
```

```python
thresh, sobelx = cv2.threshold(sobelx, 60, 255, cv2.THRESH_BINARY)
thresh, sobely = cv2.threshold(sobely, 60, 255, cv2.THRESH_BINARY)
plt.figure(figsize=[16, 9])
plt.subplot(141)
plt.title('Original')
plt.imshow(img, cmap='gray')
plt.subplot(142)
plt.title('Sobel X')
plt.imshow(sobelx, cmap='gray')
plt.subplot(143)
plt.title('Sobel Y')
plt.imshow(sobely, cmap='gray')
plt.subplot(144)
plt.title('Final')
plt.imshow(binary_sobel, cmap='gray')
```
